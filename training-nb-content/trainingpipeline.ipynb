```python
from langchain_community.document_loaders import PyMuPDFLoader

pdf_files = [
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/Blockchain-notes.pdf",
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/CC-notes.pdf",
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/CD-book.pdf",
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/CG-book.pdf",
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/CN-book.pdf",
    "C:/Users/kumar/OneDrive/Desktop/TRY-4/CSE-Core-Subject-Chatbot/Data/DSA-book.pdf"
    
]

all_docs = []

for pdf in pdf_files:
    loader = PyMuPDFLoader(pdf)
    docs = loader.load()
    all_docs.extend(docs)

print(len(all_docs))
```

# Splitting

```python
#RecursiveCharacterTextSplitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_documents(all_docs)

print(len(chunks))
```

```python
chunks[560].page_content
```

# embeddings

```python
!pip install sentence-transformers
```

```python
embedding=HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

# Vector-stores

```python
# Vector-stores : FAISS
vector_store=FAISS.from_documents(chunks,embedding)
```

```python
vector_store.save_local("vectorstore")
```

# Retrievers

```python
# Retrievers
retriever = vector_store.as_retriever(search_kwargs={"k":4})
```

```python
retriever.invoke("")
```

```python
# Augmentation
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template="""You are a helpful assistant.
Answer ONLY from the following context.
If the context is insufficient, say: I don't know.

Context:
{context}

Question:
{question}
""",
    input_variables=["context", "question"]
)
```

```python
question="What do you know about Arrays?"
```

```python
context=retriever.invoke(question)
```

```python
context
```

```python
context = " ".join(i.page_content for i in context)
```

```python

final_prompt = prompt.format(question=question, context=context)
print(final_prompt)
```
